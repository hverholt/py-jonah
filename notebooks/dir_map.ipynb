```python
from pathlib import Path

root = Path.cwd()
ignore = {".git", "__pycache__", ".ipynb_checkpoints"}
lines = [f"{root.name}/"]

def tree(folder, prefix=""):
    items = sorted(
        (p for p in folder.iterdir() if p.name not in ignore),
        key=lambda p: (p.is_file(), p.name.lower())
    )
    for i, p in enumerate(items):
        last = i == len(items) - 1
        lines.append(f"{prefix}{'└── ' if last else '├── '}{p.name}{'/' if p.is_dir() else ''}")
        if p.is_dir():
            tree(p, prefix + ("    " if last else "│   "))

tree(root)

output = "\n".join(lines)
print(output)
```
